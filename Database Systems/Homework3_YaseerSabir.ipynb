```python
#CP 2.2.2.b
import numpy as np
import copy
from math import sqrt
import numpy as np
import scipy
from scipy.optimize import fsolve
from scipy.optimize import root


def lu_no_pivoting( A_ ):
  A = copy.deepcopy(A_)
#
  n = A.shape[0]
  for k in range(0,n-1):
    if ( A[k,k] == 0. ): print("oops, zero pivot encountered\n"); break
    for i in range(k+1,n):
      A[i,k] = A[i,k] / A[k,k]
      for j in range(k+1,n):
        A[i,j] = A[i,j] - A[k,j] * A[i,k]
#
  L = np.tril(A,-1)+np.eye(n)
  U = np.triu(A)
  return L, U

def backward_substitution( A, b_ ):
  b = copy.deepcopy(b_)
  n = A.shape[0]
  for i in range(n-1,-1,-1):
    for j in range(i+1,n):
      b[i] = b[i] - A[i,j] * b[j]
    b[i] = b[i] / A[i,i]
  return b

def forward__substitution( A, b_ ):
  b = copy.deepcopy(b_)
  n = A.shape[0]
  for i in range(0,n,):
    for j in range(0,i):
      b[i] = b[i] - A[i,j] * b[j]
    b[i] = b[i] / A[i,i]
  return b

# Define A and b outside the function
A = np.array ([ [ 4., 2., 0. ],
                [ 4., 4., 2. ],
                [ 2., 2., 3. ] ] )

b = np.array ([ [ 2. ],
                [ 4. ],
                [ 6. ] ] )

# solve Ax=b with np.linalg.solve
# and check that Ax is b

x = np.linalg.solve(A, b)

print("With np.linalg.solve, we find x = \n", x)

print("\nWe check that Ax is b. Indeed [ Ax, b ] =\n", np.concatenate((A@x, b), axis=1) )

print("\nrelative backward error: || b - Ax  ||_oo / || b ||_oo =\n", np.linalg.norm(b-A@x,np.inf)/np.linalg.norm(b,np.inf) )

# using our bucket algorithms
#   `lu_no_pivoting`
#   `forward__substitution`
#   `backward_substitution`

L, U = lu_no_pivoting( A )
y = forward__substitution( L, b )
x = backward_substitution( U, y )

print("With our algorithms, we find\n")
print("y=\n",y)
print("x=\n",x)

print("\nWe check that Ax is b. Indeed [ Ax, b ] =\n", np.concatenate((A@x, b), axis=1) )

print("\nrelative backward error: || b - Ax  ||_oo / || b ||_oo =\n", np.linalg.norm(b-A@x,np.inf)/np.linalg.norm(b,np.inf) )
```

```python
import copy
def lu_no_pivoting( A_, inplace = False ):
#
  if ( inplace ):
    A = A_
  else:
    A = copy.deepcopy(A_)
# begin main code
  n = A.shape[0]
  for k in range(0,n-1):
    if ( A[k,k] == 0. ): print("oops, zero pivot encountered\n"); break
    for i in range(k+1,n):
      A[i,k] = A[i,k] / A[k,k]
      for j in range(k+1,n):
        A[i,j] = A[i,j] - A[k,j] * A[i,k]
# end main code
  if not inplace:
    L = np.tril(A,-1)+np.eye(n)
    U = np.triu(A)
    return L, U

def backward_substitution( A, b_, inplace = False ):
#
  if ( inplace ):
    b = b_
  else:
    b = copy.deepcopy(b_)
#
  n = A.shape[0]
  for i in range(n-1,-1,-1):
    for j in range(i+1,n):
      b[i] = b[i] - A[i,j] * b[j]
    b[i] = b[i] / A[i,i]
#
  if not inplace:
    return b

def forward__substitution( A, b_, inplace = False, unit = False ):
  if ( inplace ):
    b = b_
  else:
    b = copy.deepcopy(b_)
#
  n = A.shape[0]
  for i in range(0,n,):
    for j in range(0,i):
      b[i] = b[i] - A[i,j] * b[j]
    if not unit:
      b[i] = b[i] / A[i,i]
#
  if not inplace:
    return b
```

```python
#2.7.1.b
from math import sqrt
import numpy as np
import scipy
from scipy.optimize import fsolve
from scipy.optimize import root

# Define the new system:
F = lambda x: np.array([
    x[0]**2 + 4*x[1]**2 - 4,     # u^2 + 4v^2 - 4 = 0
    4*x[0]**2 + x[1]**2 - 4      # 4u^2 + v^2 - 4 = 0
])

# Two known exact solutions from symmetry (u = ±1, v = ±1)
x1 = np.array([2/np.sqrt(5), 2/np.sqrt(5)])
x2 = np.array([-2/np.sqrt(5), -2/np.sqrt(5)])

# Check backward error
print("|| F(x1) ||_oo =", f"{np.linalg.norm(F(x1),np.inf):.2E}")
print("|| F(x2) ||_oo =", f"{np.linalg.norm(F(x2),np.inf):.2E}")

# fsolve
for x, x_ in zip([np.array([1.1, 1.1]), np.array([-1.1, -1.1])], [x1, x2]):
    x = fsolve(F, x)
    print("\nx = [", f"{x[0]:+20.16f}", "]     x* = [", f"{x_[0]:+20.16f}", "]")
    print("    [", f"{x[1]:+20.16f}", "]          [", f"{x_[1]:+20.16f}", "]")
    print("forward  error: || x - x* ||_oo = ", f"{np.linalg.norm(x - x_, np.inf):7.2e}")
    print("backward error: || F(x) ||_oo   = ", f"{np.linalg.norm(F(x), np.inf):7.2e}")

# root
for x, x_ in zip([np.array([1.1, 1.1]), np.array([-1.1, -1.1])], [x1, x2]):
    x = root(F, x).x
    print("\nx = [", f"{x[0]:+20.16f}", "]     x* = [", f"{x_[0]:+20.16f}", "]")
    print("    [", f"{x[1]:+20.16f}", "]          [", f"{x_[1]:+20.16f}", "]")
    print("forward  error: || x - x* ||_oo = ", f"{np.linalg.norm(x - x_, np.inf):7.2e}")
    print("backward error: || F(x) ||_oo   = ", f"{np.linalg.norm(F(x), np.inf):7.2e}")

# Newton's Method (Jacobian)
DF = lambda x: np.array([
    [2*x[0], 8*x[1]],           # ∂/∂u and ∂/∂v of equation 1
    [8*x[0], 2*x[1]]            # ∂/∂u and ∂/∂v of equation 2
])

# Run Newton iterations
for x, x_ in zip([np.array([1.5, 1.5]), np.array([-1.5, -1.5])], [x1, x2]):
    print("** Starting Newton's method for:", x_)
    for i in range(6):
        err = np.linalg.norm(x - x_, np.inf)
        res = np.linalg.norm(F(x), np.inf)
        print(f"{i:2d}", "[", f"{x[0]:+20.16f}", f"{x[1]:+20.16f}", "]", f"{err:4.1e}", f"{res:4.1e}")
        x = x - np.linalg.solve(DF(x), F(x))

    print("\nFinal solution x =\n", x)
    print("forward  error: || x - x* ||_oo = ", f"{np.linalg.norm(x - x_, np.inf):7.2e}")
    print("backward error: || F(x) ||_oo   = ", f"{np.linalg.norm(F(x), np.inf):7.2e}")
    print()
```

```python
import numpy as np
from scipy.optimize import fsolve, root

# Define F(u, v) for the system
def F(x):
    u, v = x
    return np.array([
        u**2 - 4*v**2 - 4,
        (u-1)**2 + v**2 - 4
    ])

# Analytic Jacobian
def DF(x):
    u, v = x
    return np.array([
        [2*u, -8*v],
        [2*(u-1), 2*v]
    ])

# Exact solutions
u1 = 2.75959
v1 = 0.9507
v2 = -0.9507

x1 = np.array([u1, v1])
x2 = np.array([u1, v2])

print("|| F(x1) ||_oo =", f"{np.linalg.norm(F(x1), np.inf):.2E}")
print("|| F(x2) ||_oo =", f"{np.linalg.norm(F(x2), np.inf):.2E}")

# fsolve and root with two starting points
for x_init, x_true in zip([np.array([2.5, 1.]), np.array([2.5, -1.])], [x1, x2]):
    x_fsolve = fsolve(F, x_init)
    print("\nfsolve:")
    print("x =", x_fsolve)
    print("x* =", x_true)
    print("forward error:", f"{np.linalg.norm(x_fsolve - x_true, np.inf):.2e}")
    print("backward error:", f"{np.linalg.norm(F(x_fsolve), np.inf):.2e}")

    x_root = root(F, x_init).x
    print("root:")
    print("x =", x_root)
    print("x* =", x_true)
    print("forward error:", f"{np.linalg.norm(x_root - x_true, np.inf):.2e}")
    print("backward error:", f"{np.linalg.norm(F(x_root), np.inf):.2e}")

# Newton's method manual implementation
for x_init, x_true in zip([np.array([2.5, 1.0]), np.array([2.5, -1.0])], [x1, x2]):
    print("\nNewton's Method:")
    x = x_init.copy()
    print(f"Start: x = {x}, x* = {x_true}")
    for i in range(6):
        step = np.linalg.solve(DF(x), F(x))
        x = x - step
        print(f"Iter {i+1}: x = {x}, forward error = {np.linalg.norm(x - x_true, np.inf):.2e}, backward error = {np.linalg.norm(F(x), np.inf):.2e}")
    print("Final:")
    print("x =", x)
    print("x* =", x_true)
    print("forward error:", f"{np.linalg.norm(x - x_true, np.inf):.2e}")
    print("backward error:", f"{np.linalg.norm(F(x), np.inf):.2e}")
    print()
```

```python
#2.7.5 (b)

import numpy as np
import scipy
from scipy.optimize import fsolve

# Define the radius
R = 5.0

# Define the centers of the spheres
C1 = np.array([1., -2., 0.])
C2 = np.array([-2., 2., -1.])
C3 = np.array([4., -2., 3.])

# F(x) represents the system of equations for the three spheres
# (x - cx)^2 + (y - cy)^2 + (z - cz)^2 - R^2 = 0
F = lambda x : np.array( [
                            ( x[0] - C1[0] )**2 + ( x[1] - C1[1] )**2  + ( x[2] - C1[2] )**2  - R**2,
                            ( x[0] - C2[0] )**2 + ( x[1] - C2[1] )**2  + ( x[2] - C2[2] )**2  - R**2,
                            ( x[0] - C3[0] )**2 + ( x[1] - C3[1] )**2  + ( x[2] - C3[2] )**2  - R**2
                          ] )

# This is the Jacobian of F
DF = lambda x : np.array([
    [ 2. * ( x[0] - C1[0] ), 2. * ( x[1] - C1[1] ), 2. * ( x[2] - C1[2] ) ],
    [ 2. * ( x[0] - C2[0] ), 2. * ( x[1] - C2[1] ), 2. * ( x[2] - C2[2] ) ],
    [ 2. * ( x[0] - C3[0] ), 2. * ( x[1] - C3[1] ), 2. * ( x[2] - C3[2] ) ]
])

print("--- Using scipy.optimize.fsolve to find approximate solutions ---")

# Take an initial guess (e.g., origin)
x_guess1 = np.array( [ 0., 0., 0. ] )
sol1 = scipy.optimize.fsolve( F, x_guess1 )

print( "\nSolution 1 (from fsolve):" )
print( "x = [", f"{sol1[0]:+20.16f}", "]")
print( "    [", f"{sol1[1]:+20.16f}", "]")
print( "    [", f"{sol1[2]:+20.16f}", "]")
print( "Backward error: || F(x) ||_oo   = ", f"{np.linalg.norm(F(sol1),np.inf):7.2e}")

# Take another initial guess to find a potentially different solution
# Choose a guess far from the first solution or near where you expect another solution
x_guess2 = np.array( [ 5., 5., 5. ] ) # A different arbitrary guess
sol2 = scipy.optimize.fsolve( F, x_guess2 )

print( "\nSolution 2 (from fsolve):" )
print( "x = [", f"{sol2[0]:+20.16f}", "]")
print( "    [", f"{sol2[1]:+20.16f}", "]")
print( "    [", f"{sol2[2]:+20.16f}", "]")
print( "Backward error: || F(x) ||_oo   = ", f"{np.linalg.norm(F(sol2),np.inf):7.2e}")

# If sol1 and sol2 are very close, try another initial guess for sol2
if np.linalg.norm(sol1 - sol2, np.inf) < 1e-6:
    print("\nNote: fsolve converged to the same solution with the second guess.")
    print("Trying another initial guess for Solution 2...")
    x_guess3 = np.array([-5., -5., -5.]) # Try a more distant guess
    sol2 = scipy.optimize.fsolve(F, x_guess3)
    print("\nSolution 2 (from fsolve, new guess):")
    print( "x = [", f"{sol2[0]:+20.16f}", "]")
    print( "    [", f"{sol2[1]:+20.16f}", "]")
    print( "    [", f"{sol2[2]:+20.16f}", "]")
    print( "Backward error: || F(x) ||_oo   = ", f"{np.linalg.norm(F(sol2),np.inf):7.2e}")


# --- Multivariate Newton's Method ---
print("\n--- Applying Multivariate Newton's Method ---")

# Let's use the solutions found by fsolve as our target exact solutions for Newton's method
# (We don't know the "exact" solutions analytically for this new problem,
# so fsolve's highly accurate results serve as our reference `x_` values)
exact_sol_1 = sol1
exact_sol_2 = sol2

# Test Newton's method starting near sol1
initial_guess_newton_1 = np.array([sol1[0] - 0.2, sol1[1] + 0.1, sol1[2] + 0.1]) # Start near first center
print(f"\nNewton's Method for Solution 1 (target: {exact_sol_1})")
current_x = initial_guess_newton_1

print( f"{0:2d}", "[", f"{current_x[0]:+18.14f}", f"{current_x[1]:+18.14f}",
       f"{current_x[2]:+18.14f}",
       "]", f"{np.linalg.norm( current_x - exact_sol_1, np.inf):4.1e}",
       f"{np.linalg.norm( F(current_x), np.inf):4.1e}" )

for i in range(1, 8): # 8 iterations should be enough for quadratic convergence
    # Newton's method update
    try:
        delta_x = np.linalg.solve( DF(current_x), F(current_x) )
        current_x = current_x - delta_x
    except np.linalg.LinAlgError:
        print(f"Warning: Singular Jacobian at iteration {i}. Newton's method stopped.")
        break

    print( f"{i:2d}", "[", f"{current_x[0]:+18.14f}",
           f"{current_x[1]:+18.14f}", f"{current_x[2]:+18.14f}",
           "]", f"{np.linalg.norm( current_x - exact_sol_1, np.inf):4.1e}",
           f"{np.linalg.norm( F(current_x), np.inf):4.1e}" )

print( "\nx_final = [", f"{current_x[0]:+18.14f}", "]",
       "      x* = [", f"{exact_sol_1[0]:+18.14f}", "]")
print( "          [", f"{current_x[1]:+18.14f}", "]",
       "           [", f"{exact_sol_1[1]:+18.14f}", "]")
print( "          [", f"{current_x[2]:+18.14f}", "]",
       "           [", f"{exact_sol_1[2]:+18.14f}", "]")
print( "Forward  error: || x - x* ||_oo = ",
       f"{np.linalg.norm(current_x - exact_sol_1, np.inf):7.2e}",
       "\nBackward error: || F(x) ||_oo    = ",
       f"{np.linalg.norm(F(current_x),np.inf):7.2e}")
print("\n")


# Test Newton's method starting near sol2
initial_guess_newton_2 =  np.array([sol2[0] + 0.1, sol2[1] - 0.1, sol2[2] - 0.2]) # Start near second center
print(f"\nNewton's Method for Solution 2 (target: {exact_sol_2})")
current_x = initial_guess_newton_2

print( f"{0:2d}", "[", f"{current_x[0]:+18.14f}", f"{current_x[1]:+18.14f}",
       f"{current_x[2]:+18.14f}",
       "]", f"{np.linalg.norm( current_x - exact_sol_2, np.inf):4.1e}",
       f"{np.linalg.norm( F(current_x), np.inf):4.1e}" )

for i in range(1, 8): # 8 iterations should be enough for quadratic convergence
    # Newton's method update
    try:
        delta_x = np.linalg.solve( DF(current_x), F(current_x) )
        current_x = current_x - delta_x
    except np.linalg.LinAlgError:
        print(f"Warning: Singular Jacobian at iteration {i}. Newton's method stopped.")
        break

    print( f"{i:2d}", "[", f"{current_x[0]:+18.14f}",
           f"{current_x[1]:+18.14f}", f"{current_x[2]:+18.14f}",
           "]", f"{np.linalg.norm( current_x - exact_sol_2, np.inf):4.1e}",
           f"{np.linalg.norm( F(current_x), np.inf):4.1e}" )

print( "\nx_final = [", f"{current_x[0]:+18.14f}", "]",
       "      x* = [", f"{exact_sol_2[0]:+18.14f}", "]")
print( "          [", f"{current_x[1]:+18.14f}", "]",
       "           [", f"{exact_sol_2[1]:+18.14f}", "]")
print( "          [", f"{current_x[2]:+18.14f}", "]",
       "           [", f"{exact_sol_2[2]:+18.14f}", "]")
print( "Forward  error: || x - x* ||_oo = ",
       f"{np.linalg.norm(current_x - exact_sol_2, np.inf):7.2e}",
       "\nBackward error: || F(x) ||_oo    = ",
       f"{np.linalg.norm(F(current_x),np.inf):7.2e}")
print("\n")
```

```python
#2.7.7.b
F = lambda x : np.array([
    x[0]**2 + 4*x[1]**2 - 4,
    4*x[0]**2 + x[1]**2 - 4
])

# Per EX.2.7.3.b, we have four exact solution x1 and 2, let's trying putting in two of those
val = 2/np.sqrt(5)
x1 = np.array([+val, +val])
x2 = np.array([+val, -val])
x3 = np.array([-val, +val])
x4 = np.array([-val, -val])

# Let us check that || F(x1) ||_oo and || F(x2) ||_oo are small
print( "|| F(x1) ||_oo =", f"{np.linalg.norm(F(x1),np.inf):.2E}" )
print( "|| F(x2) ||_oo =", f"{np.linalg.norm(F(x2),np.inf):.2E}" )
print( "|| F(x3) ||_oo =", f"{np.linalg.norm(F(x3),np.inf):.2E}" )
print( "|| F(x4) ||_oo =", f"{np.linalg.norm(F(x4),np.inf):.2E}" )


for x, x_ in zip( [ np.array( [ 0.9, 0.9 ] ), np.array( [ 0.9, -0.9 ] ), np.array([-0.9, 0.9]), np.array([-0.9, -0.9])],
                  [ x1, x2, x3, x4 ] ):
  A = np.eye(2,2)

  print( f"{0:2d}", "[", f"{x[0]:+20.16f}", f"{x[1]:+20.16f}",\
        "]", f"{np.linalg.norm( x - x_, np.inf):4.1e}", \
        f"{np.linalg.norm( F(x), np.inf):4.1e}" )
  tol = 1e-12
  for i in range(1,12):

    x0 = x
    x = x - np.linalg.solve( A, F(x) )
    Delta = ( F(x) - F(x0) )
    delta = x - x0
    if np.linalg.norm(F(x), np.inf) < tol or np.linalg.norm(delta) < tol:
      break                          # we’re converged – stop iterating

    A = A + np.outer ( ( Delta - ( A @ delta ) ), delta / (delta.T @ delta) )
    denom = delta.T @ delta
    if np.abs(denom) < 1e-14:          # step is (numerically) zero – skip update
      continue


    print( f"{i:2d}", "[", f"{x[0]:+20.16f}", f"{x[1]:+20.16f}",\
            "]", f"{np.linalg.norm( x - x_, np.inf):4.1e}",\
            f"{np.linalg.norm( F(x), np.inf):4.1e}" )

  print( "\nx = [", f"{x[0]:+20.16f}", "]"\
          "     x* = [", f"{x_[0]:+20.16f}", "]")
  print( "    [", f"{x[1]:+20.16f}", "]"\
          "          [", f"{x_[1]:+20.16f}", "]")
  print( "forward  error: || x - x* ||_oo = ", f"{np.linalg.norm(x-x_,np.inf):7.2e}",\
          "\nbackward error: || F(x) ||_oo   = ", f"{np.linalg.norm(F(x),np.inf):7.2e}")
  print("\n")
```
